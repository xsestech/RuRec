## Algorithm example

```python
player_starts_train['item_id'].value_counts()
```

```python
# %%timeit 

user_id = 24477183

def process(user_id):
    players_user_ids = torch.tensor(numpy.array(player_starts_train['user_id']), device=backend)
    players_item_ids = torch.tensor(numpy.array(player_starts_train['item_id']), device=backend)
    players = torch.stack((players_user_ids, players_item_ids), dim=1)
    
    rec_players = players[players[:, 0] == user_id][:, 1].unique()[-5:]
    rec_players = torch.tensor([170129])
    
    rec_matches = torch.rand(1, 2, device=backend)
    for rec_plr in rec_players:
        rec_matches = torch.cat((rec_matches, players[players[:, 1] == rec_plr]), dim=0)
    rec_matches = rec_matches[1:]
    matches_unique_users = rec_matches[:, 0].unique().int()
    rec_users = matches_unique_users[torch.randint(0, matches_unique_users.size()[0], (1000,))]
    rec_users = torch.stack((rec_users, torch.tensor([(rec_matches[:, 0] == rec_user).int().sum() for rec_user in rec_users], device=backend)), dim=1)
    sorted_rec_users_ids = rec_users.sort(0)[1][:, 1]
    rec_users = rec_users[sorted_rec_users_ids.flip(0)][:100]
    rec_users_vectors = torch.rand(1, 5, device=backend)
    for rec_user in rec_users:
        vector = rec_matches[rec_matches[:, 0] == rec_user[0]][:, 1].unique()
        # print(vector)
        bow = torch.zeros(5, device=backend)
        for i in range(len(rec_players)):
            bow[i] = rec_players[i] in vector
        rec_users_vectors = torch.cat((rec_users_vectors, bow[None]))
    rec_users_vectors = rec_users_vectors[1:]
    rec_users_vectors_sums = rec_users_vectors.sum(1)
    # print(rec_users_vectors_sums)
    _, ids_sorted_by_in = torch.sort(rec_users_vectors_sums)
    ids_sorted_by_in = ids_sorted_by_in.flip(0)
    rec_users = rec_users[ids_sorted_by_in]
    
    allu_videos = torch.tensor([], device=backend)
    for rec_user in rec_users:
        videos = players[players[:, 0] == rec_user[0]][:, 1].unique()
        videos
        allu_videos = torch.cat((allu_videos, videos), dim=0)
    uallu_videos = allu_videos.unique()
    videos_cnt = torch.tensor([(allu_videos == video).int().sum() for video in uallu_videos], device=backend)
    _, ids = torch.sort(videos_cnt)
    ids = ids.flip(0)
    videos = uallu_videos[ids].int()
    videos = videos[reduce(lambda x, y: x * y, [videos != video for video in rec_players])]
    videos = list(map(lambda x: f'video_{x}', videos[:10].tolist()))
    return str(videos)

print(process(user_id))
```
